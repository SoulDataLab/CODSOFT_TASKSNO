```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

# Agar file nahi hai, toh yahan se upload kar le
uploaded = files.upload()

# Cleaned dataset load karna
file_path = "cleaned_marketing_campaign.csv"
df = pd.read_csv(file_path)

sns.set_theme(style="whitegrid")
print("Dataset loaded successfully for Task 5!")
df.head(3)
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Education', hue='Response', palette='Set2')
plt.title('Campaign Response by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Count')
plt.legend(title='Response')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Marital_Status', hue='Response', palette='Set3')
plt.title('Campaign Response by Marital Status')
plt.xlabel('Marital Status')
plt.ylabel('Count')
plt.legend(title='Response')
plt.show()
```

```python
spending_cols = ['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']
spending_sums = df[spending_cols].sum()

plt.figure(figsize=(10, 5))
spending_sums.plot(kind='bar', color='coral', edgecolor='black')
plt.title('Total Customer Spending Across Product Categories')
plt.xlabel('Product Categories')
plt.ylabel('Total Amount Spent')
plt.xticks(rotation=45)
plt.show()
```

```python
summary_report = pd.DataFrame({
    'Category': spending_cols,
    'Total_Revenue': spending_sums.values
})

summary_report.to_csv("task5_visual_summary_report.csv", index=False)
print("Task 5 visual summary report successfully saved as 'task5_visual_summary_report.csv'.")
```
